```python
from sklearn.datasets import make_blobs
X,y = make_blobs(n_samples=500, n_features=2, centers=4,cluster_std=2.0, random_state=42)
print(X[:10])
print(y[:10])
import pandas as pd
import seaborn as sns
df = pd.DataFrame(dict(x=X[:,0] , y=X[:,1], label = y))
sns.scatterplot(x=df.x, y=df.y,data=df,hue=df.label,palette='bright')
```

```python
#Defeine (sequential)
from keras.models import Sequential
from keras.layers import Dense, Input
model = Sequential()
model.add(Input(shape = (3,)))
model.add(Dense(2, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
model.summary()
```

```python
#Defeine (functional)
from keras.models import Model
from keras.layers import Dense, Input
input_layer = Input(shape = (3,))
hidden_layer = Dense(2, activation='relu')(input_layer)
output_layer = Dense(1, activation='sigmoid')(hidden_layer)
model = Model(inputs=input_layer, outputs=output_layer)
model.summary()
```

```python
from keras.utils import plot_model
plot_model(model,to_file='functional1.png',show_shapes=True)
```

```python
from keras.models import Model
from keras.layers import Input, Dense, Concatenate
# Input layer
inputs = Input(shape=(5,))
x1 = Dense(20, activation='tanh')(inputs)
x1 = Dense(10, activation='tanh')(x1)
x1 = Dense(8, activation='tanh')(x1)
x1 = Dense(4, activation='tanh')(x1)

x2 = Dense(10, activation='relu')(inputs)
x2 = Dense(10, activation='relu')(x2)
x2 = Dense(5, activation='relu')(x2)
merged = Concatenate()([x1, x2])


x = Dense(20, activation='relu')(merged)
outputs = Dense(3, activation='softmax')(x)
model = Model(inputs=inputs, outputs=outputs)
model.summary()
plot_model(model,to_file='func.png',show_shapes=True)
```

```python
from keras.models import Model
from keras.layers import Input, Dense, Concatenate

# Input layer
inputs = Input(shape=(2,))

# ฝั่งซ้าย (Left Branch)
x1 = Dense(16, activation='relu')(inputs)
x1 = Dense(8, activation='relu')(x1)

# ฝั่งขวา (Right Branch)
x2 = Dense(12, activation='relu')(inputs)
x2 = Dense(6, activation='relu')(x2)

merged = Concatenate()([x1, x2])

# เลเยอร์ด้านล่างหลังจากรวมกันแล้ว
x = Dense(10, activation='relu')(merged)
x = Dense(5, activation='relu')(x)
outputs = Dense(4, activation='softmax')(x)

# สร้างและสรุปโมเดล
model = Model(inputs=inputs, outputs=outputs)
model.summary()
plot_model(model, to_file='func.png', show_shapes=True)
```
